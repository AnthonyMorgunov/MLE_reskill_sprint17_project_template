# Задание 2. Сегментация коммерческих запросов

Компания «Экзосфера» хочет выделить профили коммерческого спроса. Выполните исследование по разделам ниже. Метрики кластеризации считайте в полном масштабированном пространстве.

## SQL-запрос и выгрузка

Сформулируйте бизнес-вопрос и подтвердите, что одна строка соответствует `request_id`. Вставьте итоговый SQL как Markdown-блок `запросы`. Опишите агрегацию территорий и размер выгрузки.

## Аудит, выбор признаков и масштабирование

Проверьте пропуски, дубли, распределения и выбросы. Отделите числовые признаки от категорий для постфактум-интерпретации. Обработайте пропуски и масштабируйте признаки.

## K-Means: локоть и Silhouette

Исследуйте диапазон K методом локтя и Silhouette. Метрики считайте в полном масштабированном пространстве.

## DBSCAN: k-distance, параметры и шум

Постройте k-distance-график, обоснуйте `eps` и `min_samples`, посчитайте кластеры и шум. Не удаляйте `-1` молча.

## PCA и UMAP

После кластеризации проанализируйте `explained_variance_ratio_`, постройте две воспроизводимые проекции и раскрасьте их по меткам каждого алгоритма. Сравните видимые группы, шум и ограничения проекций.

## Профилирование в исходных единицах

Для каждой группы покажите размер, среднее, медиану и отличающие признаки в исходных единицах. Шум опишите отдельно.

## Бизнес-рекомендация и ограничения

Сопоставьте метрики, визуализации, охват и бизнес-смысл. Обсудите зависимость решения от параметров и масштаба; если устойчивость не проверялась отдельно, явно назовите это ограничение, а не подтверждённое свойство. Не выбирайте алгоритм только по максимальному Silhouette. Выберите K-Means или DBSCAN, свяжите продуктовые пакеты с конкретными профилями и перечислите ограничения.
