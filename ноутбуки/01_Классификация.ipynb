# Задание 1. Прогноз качества наблюдения

Компания «Экзосфера» хочет заранее оценивать риск непрохождения контроля качества наблюдения. Выполните исследование по разделам ниже и сохраните результаты и выводы.

## SQL-запрос и выгрузка

Сформулируйте момент прогноза и цену FP и FN. Вставьте итоговый SQL как Markdown-блок `запросы`. Укажите единицу анализа, таблицы, размер выгрузки и локальный путь.

## Аудит данных, идентификатор и дубли

Загрузите CSV. Проверьте форму, типы, пропуски, баланс классов и дубли без `session_id`. Объясните роль ID. Обоснуйте обработку содержательных копий до разделения; новый ID сам по себе не делает копию независимым наблюдением. Проверьте, что копии одного наблюдения не попали в разные выборки.

## Разделение, инженерия признаков и безопасная предобработка

Выберите пропорции обучающей, валидационной и тестовой выборок и зафиксируйте `random_state`. Объясните допущение стационарности. Создайте два или три признака. Соберите `Pipeline` и `ColumnTransformer`, обучая преобразования только на обучающей части. Учтите особую роль отношений параметров для SVM.

## Сравнение моделей

Обучите логистическую регрессию с обоснованными `penalty` и `C`, одну нелинейную модель и один изученный бустинг. Обоснуйте решение о весах классов. Сравните Precision, Recall, F1 и матрицы ошибок на валидационной выборке.

## Optuna для лучшего алгоритма сравнения

Настройте лучший алгоритм за 20–30 испытаний со стратифицированной кросс-валидацией. При практически равных результатах заранее сформулируйте и обоснуйте правило выбора по валидационной выборке и требованиям бизнеса. Не используйте тестовую выборку.

## Финальная оценка на тестовой выборке

Один раз оцените финальную конфигурацию на тестовой выборке. Покажите метрики и проверьте `F1 >= 0.75`.

## SHAP-интерпретация итоговой модели после Optuna

SHAP интерпретацию произведите для лучшего алгоритма. Скорее всего, это будет алгоритм на основе деревьев: случайный лес или бустинг.

После настройки выбранного алгоритма с помощью Optuna выполните SHAP-интерпретацию его итоговой настроенной модели. Укажите название модели и итоговый F1 на тестовой выборке. Интерпретируйте именно итоговую модель, даже если она не основана на деревьях.

Покажите основные факторы, связанные с предсказаниями. Выводы SHAP объясняют работу модели и не доказывают причинное влияние признаков на качество снимка.
